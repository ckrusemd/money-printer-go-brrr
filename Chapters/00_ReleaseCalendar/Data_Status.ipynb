

Essential missing inputs stop publication and leave the previous book available.
An optional series that could not be retrieved is marked unavailable and excluded
from this build. Historical comparisons use the dates and data preparation
described in their chapters.

Use **Find a series** to search by provider, request identifier, or availability.
Scroll the table horizontally on narrow screens to see all columns.


```r
#| echo: false
#| warning: false
#| label: source-status
#| output: asis
options(repr.plot.width = 10, repr.plot.height = 7, repr.plot.res = 120)
source_status <- book_source_status()
if (nrow(source_status)) {
  options(book.figure.alt = "Data freshness table: search by provider, request identifier or availability")
  status_table <- DT::datatable(
    source_status,
    rownames = FALSE,
    options = list(
      scrollX = TRUE, scrollY = "300px", scrollCollapse = TRUE,
      pageLength = 10, lengthMenu = c(10, 25, 50),
      language = list(search = "Find a series:",
                      lengthMenu = "Show _MENU_ requests",
                      info = "Showing _START_ to _END_ of _TOTAL_ requests",
                      infoFiltered = "(filtered from _MAX_ total requests)",
                      zeroRecords = "No matching requests")
    )
  )
  status_table$sizingPolicy$browser$padding <- 0
  status_table <- htmlwidgets::prependContent(status_table, htmltools::tags$style(
    "body { margin: 0; font: 15px/1.4 system-ui, sans-serif; color: #343a40; }
     table { font-variant-numeric: tabular-nums; }
     input, select { font: inherit; }
     :focus-visible { outline: 3px solid #005a9c; outline-offset: 2px; }
     .dataTables_wrapper .dataTables_paginate .paginate_button { padding: .4em .6em; }"
  ))
  IRdisplay::display_html(repr_html.htmlwidget(status_table))
} else {
  cat("No remote observations were requested in this build.\n")
}
book_write_manifest()
```
