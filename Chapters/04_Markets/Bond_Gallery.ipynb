# Bond ETF chart gallery

Explore bond ETF price histories, moving averages, momentum, and comparisons with inflation and commodities. The groups below distinguish short-maturity Treasury holdings, inverse exposure, and other long bond exposure; use each fund's name and ticker to identify its mandate. Return to the [market snapshots](equity-markets-and-the-sp-500.html) for a compact Treasury-bond view.

Each chart follows the same order as the [market gallery](market-chart-gallery.html): price history, recent prices, indicator heatmap, moving-average distance, RSI, currency and commodity comparisons, then rebased nominal and inflation-adjusted prices. These are opening-price histories, not total-return series with reinvested distributions. Short and leveraged funds can have very different price paths from ordinary bond holdings.

The heatmap's distribution scores are historical comparisons, not forecasts. Non-trading days carry the previous price forward; missing symbols are marked unavailable, and panels requiring longer histories may be incomplete.
