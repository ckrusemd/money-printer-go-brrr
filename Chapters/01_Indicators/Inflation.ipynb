## Consumer price changes

Denmark’s current consumer CPI replaces the archived PRIS111 table after the provider’s rebasing. The earlier PRIS114 net price index excludes parts of consumer taxation and is not the household purchasing-power deflator. The main US series is unadjusted consumer CPI; the seasonally adjusted US series is separately labelled. Rebase indices before comparing their levels.


```r
#| echo: false
#| warning: false
research_chart(ctx$data,c("us_cpi","dk_cpi"),"Consumer inflation · annual changes",transform="growth")
```

```r
#| echo: false
#| warning: false
research_chart(ctx$data,c("us_cpi","dk_cpi"),"Monthly CPI changes · not seasonally adjusted",transform="growth",lag=1L)
```

## Real asset growth

Choose real DKK to convert prices with contemporaneous exchange rates and then divide by Danish consumer CPI. Domestic views use the currency’s consumer index. Real values and nominal values are indexed to a common historical starting window. A change in the price level is different from a change in the inflation rate.


```r
#| echo: false
#| warning: false
research_value_chart(ctx$data,c("sp500","c25","gold","bitcoin"),"Assets: nominal prices versus purchasing power")
```

## Wages and cash

US hourly earnings cover private-sector employees and are not household disposable income. The wage panel shows purchasing-power growth. The cash panel holds the nominal amount fixed; it excludes deposit interest, tax and fees.


```r
#| echo: false
#| warning: false
research_value_chart(ctx$data,"us_wage","US nominal and real hourly earnings")
```

```r
#| echo: false
#| warning: false
cash <- research_series(ctx$data,"dk_cpi"); cash$value <- 100*cash$value[1]/cash$value
p <- ggplot2::ggplot(cash,ggplot2::aes(date,value))+ggplot2::geom_line()+theme_money_printer_go_brrr(11)+ggplot2::labs(title="Purchasing power of fixed nominal DKK cash",y="Initial purchasing power=100",x=NULL)
plotly::ggplotly(p)
```
