# Market chart gallery

Compare broad equity indices, Bitcoin, and precious metals using the same indicator panels. For a quicker view, return to the [six market snapshots](equity-markets-and-the-sp-500.html). Values are derived from Yahoo Finance, FRED, and Statistics Denmark data available at build time.

Read each panel from top to bottom: full price history and the past year, an indicator heatmap, the distance from the 200-observation moving average, RSI, currency and commodity comparisons, and nominal versus inflation-adjusted prices rebased to January 2015. A rebasing panel requires history covering that base date. Dashed support lines, where present, extend fits through the selected historical dates; they are not forecasts.

The heatmap labels give both indicator values and historical distribution scores. The RSI panel uses 30 observations; the heatmap uses its separately specified indicator settings. Prices are carried forward over non-trading days. Missing symbols and insufficient history are marked unavailable; currency and commodity panels may have gaps where matching observations are absent.

## S&P 500 {#gallery-sp500}


These galleries retain the earlier exploratory preparation, including opening-price and calendar-filled display series. The [main equity chapter](equity-markets-and-the-sp-500.html) uses completed closes and native sessions for current return statistics and forecast comparisons. Gallery transformations are not model-training inputs.
