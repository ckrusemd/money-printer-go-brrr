## Read the evidence

A benchmark remains the published model unless an alternative improves held-out MAE by at least 5%, has at least 30 matured forecast origins, and passes a block-bootstrap skill check. Revised economic data limit historical realism; each model table identifies that basis. Missing forecast bands mean calibration evidence is insufficient.

Index charts show price growth excluding dividends. The world series is an ACWI ETF price proxy, and gold is a futures-price reference. These are different instruments, not interchangeable investments.
