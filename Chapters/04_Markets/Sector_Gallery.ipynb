# Sector ETF chart gallery

Compare sector ETFs using a common set of price and indicator panels. Long exposures appear first, followed by short and leveraged funds. Return to the [market snapshots](equity-markets-and-the-sp-500.html) for broad-index context, or the [market gallery](market-chart-gallery.html) for the panel reading guide.

Read downward through price history, recent prices, the indicator heatmap, moving-average distance, RSI, currency and commodity comparisons, and rebased nominal and inflation-adjusted prices. The charts use opening prices, and do not include reinvested distributions. Historical heatmap scores describe relative indicator levels, not the probability of a future return.

Compare funds within their stated mandate: a short or leveraged fund's history does not represent an ordinary long investment in that sector. Non-trading days carry the previous price forward; unavailable symbols and insufficient histories are identified in the charts.
