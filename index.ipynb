## Sources and freshness

The compilation time records when this edition was built. It does not mean every series has an observation from that day. Economic releases can lag their reference month or quarter, historical observations can be revised, and market holidays create gaps in trading data. Where a chart carries the last value forward, that value is an earlier observation rather than a new trade or release.

Check [this edition's data status](data-status.html) for each provider's retrieval time and latest observation date, together with any failed optional requests. These dates distinguish when the book retrieved a series from how recent its underlying observations are.

FRED supplies US economic series; Statistics Denmark supplies Danish statistics; Yahoo Finance supplies market histories; and Finans Danmark supplies mortgage-rate data. Follow the source links above for release notes and definitions. The Danish fund-tax tables use dated source snapshots, with their provenance described in the [ETF chapter](etf-tracking-and-analysis.html#danish-fund-tax).

Builds retrieve essential source data before reusing derived calculations. A failed essential retrieval must stop publication so a newly dated edition cannot silently present an old cached source response as fresh data. Optional unavailable market symbols are identified in the data-status appendix. Cached calculations reuse matching inputs; they do not make those inputs more recent.

The analysis is historical and exploratory. Read the observation dates, units, and chapter methods alongside each chart; indicator scores and fitted relationships are not forecasts or probabilities of future returns. The [publication source](https://github.com/ckrusemd/money-printer-go-brrr) contains the chapter code, dependency lockfile, and build checks.
