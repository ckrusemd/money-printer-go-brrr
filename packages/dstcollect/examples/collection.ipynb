# Collecting Statistics Denmark data

Discover a table, inspect its codes, and collect observations with provenance. The package preserves native frequency and explicit category meanings.


```r
source("scripts/notebooks/bootstrap.R")
library(dstcollect)
```

```r
meta <- dst_metadata("FOLK1A")
meta$dimensions
subset(meta$categories, dimension_id == "OMRÅDE" & category_code %in% c("000", "101"))
```

```r
population <- dst_collect("FOLK1A", list(OMRÅDE=c("000", "101"), KØN="TOT", ALDER="IALT", CIVILSTAND="TOT", Tid="(1)"), metadata=meta)
population
dst_as_tibble(population, dimensions="labels", concepts=TRUE)
```

```r
prices <- dst_collect("PRIS90", list(BOLTYP="0111", ENHED=c("100", "210"), Tid="(1)"))
combined <- dst_bind(population, prices)
combined$data
combined$concepts
```

Missing values remain missing. Series IDs use source codes, not labels. Percentage changes and index levels retain distinct measures and units. Unmapped categories remain available under native codes.
