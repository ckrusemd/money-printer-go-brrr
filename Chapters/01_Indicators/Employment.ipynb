## Jobs and employed people

US payroll employment counts nonfarm jobs, while the household survey counts employed people. Danish employee statistics cover a different population and industrial classification. Levels are shown with units and should not be ranked as though the countries had equal population or identical definitions. The current Danish industry-classification table replaces the discontinued series.

## Unemployment and participation

Danish quarterly figures are labour-force-survey measures, not registered unemployment. US rates use a civilian age-16+ population; Danish labour-force coverage differs. Changes in employment, unemployment and participation can move differently because people enter or leave the labour force.


```r
#| echo: false
#| warning: false
research_chart(ctx$data,c("us_unemployment","dk_rates_LPCT"),"Labour-force unemployment rates · definitions differ")
```

```r
#| echo: false
#| warning: false
research_chart(ctx$data,c("us_participation","dk_rates_EFK","dk_rates_BFK"),"Participation and Danish employment rate")
```
